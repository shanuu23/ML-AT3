### B.1 Explore Dataset

```python
df.head()
```

```python
# Basic dataset information
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame(name="data_type"))

print("\nFirst 5 rows:")
display(df.head())

print("\nSummary statistics for numerical columns:")
display(df.describe(include=[np.number]))

print("\nSummary statistics for categorical columns:")
display(df.describe(include=["object"]))
```

```python
# Missing value analysis
missing_summary = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": (df.isna().mean() * 100).round(2),
    "non_missing_count": df.notna().sum()
}).sort_values(by="missing_percentage", ascending=False)

missing_summary
```

```python
# Duplicate analysis
total_rows = len(df)
duplicate_rows = df.duplicated().sum()
unique_person_ids = df["person_id"].nunique()
duplicate_person_ids = total_rows - unique_person_ids

duplicate_summary = pd.DataFrame({
    "metric": [
        "Total rows",
        "Fully duplicated rows",
        "Unique person_id values",
        "Rows beyond unique person_id count"
    ],
    "value": [
        total_rows,
        duplicate_rows,
        unique_person_ids,
        duplicate_person_ids
    ]
})

duplicate_summary
```

```python
# Check repeated person_id values
person_id_duplicates = (
    df["person_id"]
    .value_counts()
    .reset_index()
)

person_id_duplicates.columns = ["person_id", "count"]

person_id_duplicates.head(10)
```

```python
# Visualise missing percentage by column
missing_chart_data = missing_summary.reset_index().rename(columns={"index": "column"})

alt.Chart(missing_chart_data).mark_bar().encode(
    x=alt.X("missing_percentage:Q", title="Missing Percentage (%)"),
    y=alt.Y("column:N", sort="-x", title="Column"),
    tooltip=["column", "missing_count", "missing_percentage"]
).properties(
    title="Missing Values by Column",
    width=700,
    height=350
)
```

```python
dataset_insights = """
The person dataset contains 25,163 rows and 10 columns. The main identifier is person_id, which can be used later to join this dataset with customer, sales, or order-related datasets. The dataset contains personal information such as name fields, person type, email promotion preference, and additional contact information.
The dataset has an important duplicate issue. There are 5,191 fully duplicated rows, and only 19,972 unique person_id values. This means duplicate records should be removed before using this dataset for modelling or joining with other datasets. If duplicates are not removed, the same person may be counted multiple times, which can distort customer-level analysis.
The most complete columns are person_id, first_name, and last_name. However, name-related fields are not very useful for a regression model because they do not provide meaningful business behaviour information. Several columns have high missing values: suffix, additional_contact_info, title, and middle_name. These columns are unlikely to be useful for predicting sales amount and should probably be removed during data preparation.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='dataset_insights', value=dataset_insights)
```

### B.2 Explore Feature of Interest `person_type`

```python
# Explore person_type distribution
person_type_summary = (
    df["person_type"]
    .value_counts(dropna=False)
    .reset_index()
)

person_type_summary.columns = ["person_type", "count"]
person_type_summary["percentage"] = (
    person_type_summary["count"] / len(df) * 100
).round(2)

person_type_summary
```

```python
# Replace missing values for visualisation only
person_type_chart_data = df.copy()
person_type_chart_data["person_type"] = person_type_chart_data["person_type"].fillna("Missing")

person_type_counts = (
    person_type_chart_data["person_type"]
    .value_counts()
    .reset_index()
)

person_type_counts.columns = ["person_type", "count"]

alt.Chart(person_type_counts).mark_bar().encode(
    x=alt.X("person_type:N", title="Person Type", sort="-y"),
    y=alt.Y("count:Q", title="Number of Records"),
    tooltip=["person_type", "count"]
).properties(
    title="Distribution of Person Type",
    width=600,
    height=350
)
```

```python
feature_1_insights = """
The person_type feature categorises each person into a type. The most common value is IN, which accounts for the vast majority of records. Other person types such as SC, GC, EM, VC, and SP appear much less frequently. This shows that the dataset is highly imbalanced across person types.
For the regression use case, person_type may be useful because different person categories may have different purchasing behaviour or business roles. For example, some person types may represent individual customers, store contacts, employees, vendor contacts, or sales-related contacts. If this dataset is joined with customer and sales transaction data, person_type could help the model learn whether certain categories of people are associated with higher or lower order values.
There are a small number of missing values in person_type. These can be handled by filling them with 'Unknown' during data preparation. Since person_type is categorical, it should be encoded using one-hot encoding or another categorical encoding method before training regression models.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `email_promotion`

```python
# Explore email_promotion distribution
email_promotion_summary = (
    df["email_promotion"]
    .value_counts(dropna=False)
    .reset_index()
)

email_promotion_summary.columns = ["email_promotion", "count"]
email_promotion_summary["percentage"] = (
    email_promotion_summary["count"] / len(df) * 100
).round(2)

email_promotion_summary
```

```python
# Prepare email_promotion for visualisation
email_chart_data = df.copy()
email_chart_data["email_promotion"] = email_chart_data["email_promotion"].fillna("Missing")
email_chart_data["email_promotion"] = email_chart_data["email_promotion"].astype(str)

email_promotion_counts = (
    email_chart_data["email_promotion"]
    .value_counts()
    .reset_index()
)

email_promotion_counts.columns = ["email_promotion", "count"]

alt.Chart(email_promotion_counts).mark_bar().encode(
    x=alt.X("email_promotion:N", title="Email Promotion Category", sort="-y"),
    y=alt.Y("count:Q", title="Number of Records"),
    tooltip=["email_promotion", "count"]
).properties(
    title="Distribution of Email Promotion Preference",
    width=600,
    height=350
)
```

```python
feature_2_insights = """
The email_promotion feature shows the promotional email preference of each person. The values are mainly 0, 1, and 2, with 0 being the most common category. This suggests that many people either do not receive promotional emails or belong to the lowest promotional email category.
This feature may be useful for the regression use case because customers who receive or accept promotional emails may have different purchasing behaviour compared with customers who do not. For example, customers with higher email promotion values may be more responsive to campaigns, offers, or discounts, which could affect order value or total sales amount.
The feature has a small number of missing values. These can be filled using the most frequent value or treated as a separate 'Unknown' category. Since email_promotion is represented using numeric codes but behaves like a category, it should be treated as a categorical variable during modelling rather than as a continuous numerical variable.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `name_style`
