```python
header_clean["order_date"] = pd.to_datetime(
    header_clean["order_date"],
    errors="coerce"
)

header_prepared_df = header_clean[
    [
        "sales_order_id",
        "customer_id",
        "territory_id",
        "order_date",
        "online_order_flag",
        "status"
    ]
].copy()

header_prepared_df = header_prepared_df.drop_duplicates(subset=["sales_order_id"])

header_prepared_df["territory_id"] = header_prepared_df["territory_id"].fillna("Unknown")
header_prepared_df["status"] = header_prepared_df["status"].fillna("Unknown")
header_prepared_df["online_order_flag"] = header_prepared_df["online_order_flag"].fillna(
    header_prepared_df["online_order_flag"].mode()[0]
)

header_prepared_df["order_year"] = header_prepared_df["order_date"].dt.year
header_prepared_df["order_month"] = header_prepared_df["order_date"].dt.month
header_prepared_df["order_dayofweek"] = header_prepared_df["order_date"].dt.dayofweek
header_prepared_df["is_weekend"] = header_prepared_df["order_dayofweek"].isin([5, 6]).astype(int)

for col in ["order_year", "order_month", "order_dayofweek"]:
    header_prepared_df[col] = header_prepared_df[col].fillna(
        header_prepared_df[col].mode()[0]
    )

header_prepared_df["is_weekend"] = header_prepared_df["is_weekend"].fillna(0)

header_prepared_df["territory_id"] = header_prepared_df["territory_id"].astype(str)
header_prepared_df["status"] = header_prepared_df["status"].astype(str)
header_prepared_df["online_order_flag"] = header_prepared_df["online_order_flag"].astype(str)

header_prepared_df.head()
```

### B.n Fixing "Prepare Sales Order Detail"

> You can add more cells related to other issues in this section

```python
numeric_detail_columns = [
    "sales_order_detail_id",
    "order_quantity",
    "unit_price",
    "unit_price_discount",
    "line_total"
]

for col in numeric_detail_columns:
    detail_clean[col] = pd.to_numeric(detail_clean[col], errors="coerce")

numeric_conversion_check = pd.DataFrame({
    "column": numeric_detail_columns,
    "missing_after_conversion": [
        detail_clean[col].isna().sum() for col in numeric_detail_columns
    ]
})

numeric_conversion_check
```

```python
rows_before_invalid_filter = len(detail_clean)

detail_clean = detail_clean[
    (detail_clean["order_quantity"] > 0) &
    (detail_clean["unit_price"] > 0) &
    (detail_clean["unit_price_discount"] >= 0) &
    (detail_clean["unit_price_discount"] <= 1) &
    (detail_clean["line_total"] > 0)
].copy()

rows_after_invalid_filter = len(detail_clean)

invalid_transaction_summary = pd.DataFrame({
    "metric": [
        "Rows before invalid transaction filtering",
        "Rows after invalid transaction filtering",
        "Rows removed"
    ],
    "value": [
        rows_before_invalid_filter,
        rows_after_invalid_filter,
        rows_before_invalid_filter - rows_after_invalid_filter
    ]
})

invalid_transaction_summary
```

```python
order_detail_agg = (
    detail_clean
    .groupby("sales_order_id")
    .agg(
        order_total=("line_total", "sum"),
        number_of_line_items=("sales_order_detail_id", "count"),
        total_quantity=("order_quantity", "sum"),
        average_quantity=("order_quantity", "mean"),
        average_unit_price=("unit_price", "mean"),
        max_unit_price=("unit_price", "max"),
        min_unit_price=("unit_price", "min"),
        std_unit_price=("unit_price", "std"),
        average_discount=("unit_price_discount", "mean"),
        max_discount=("unit_price_discount", "max"),
        discounted_line_count=("unit_price_discount", lambda x: (x > 0).sum()),
        unique_products=("product_id", "nunique"),
        unique_special_offers=("special_offer_id", "nunique")
    )
    .reset_index()
)

order_detail_agg["std_unit_price"] = order_detail_agg["std_unit_price"].fillna(0)

print("Order-level detail dataset shape:", order_detail_agg.shape)
order_detail_agg.head()
```

Joining Datasets

```python
model_df = order_detail_agg.merge(
    header_prepared_df,
    on="sales_order_id",
    how="left",
    suffixes=("", "_header")
)

model_df = model_df.merge(
    customer_prepared_df,
    on="customer_id",
    how="left",
    suffixes=("", "_customer")
)

model_df = model_df.merge(
    person_prepared_df,
    on="person_id",
    how="left"
)

print("Joined modelling dataset shape:", model_df.shape)
model_df.head()
```

```python
if "territory_id_customer" in model_df.columns:
    model_df["territory_id"] = model_df["territory_id"].replace("nan", np.nan)
    model_df["territory_id_customer"] = model_df["territory_id_customer"].replace("nan", np.nan)

    model_df["territory_id"] = model_df["territory_id"].fillna(
        model_df["territory_id_customer"]
    )

    model_df = model_df.drop(columns=["territory_id_customer"])

model_df["territory_id"] = model_df["territory_id"].fillna("Unknown").astype(str)
model_df["person_type"] = model_df["person_type"].fillna("Unknown").astype(str)
model_df["email_promotion"] = model_df["email_promotion"].fillna("Unknown").astype(str)
model_df["status"] = model_df["status"].fillna("Unknown").astype(str)
model_df["online_order_flag"] = model_df["online_order_flag"].fillna("Unknown").astype(str)

model_df.head()
```
