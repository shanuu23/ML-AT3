### B.3 Explore Feature of Interest `territory_id              `

```python
len(df['territory_id'].unique())
```

```python
df['territory_id'].value_counts()
```

```python
df_avg_due = df.groupby('territory_id')['total_due'].mean().reset_index()
df_avg_due_top10 = df_avg_due.sort_values(by='total_due', ascending=False).head(10)

# Create a mapping from original territory_id to numerical IDs
territory_id_mapping = {id: i + 1 for i, id in enumerate(df_avg_due_top10['territory_id'])}

# Apply the mapping to the DataFrame
df_avg_due_top10['territory_numerical_id'] = df_avg_due_top10['territory_id'].map(territory_id_mapping)

display(df_avg_due_top10.head(10))
```

```python
# Chart for Average Total Due (Bars)
bars = alt.Chart(df_avg_due_top10).mark_bar(opacity=0.7, color='lightblue').encode(
    x=alt.X('territory_numerical_id:O', title='Territory ID (Numerical)'),
    y=alt.Y('total_due', axis=alt.Axis(title='Average Total Due', titleColor='lightblue')),
    tooltip=['territory_id', 'territory_numerical_id', 'total_due']
)

# Filter the original DataFrame to include only the top 10 territory_ids
top10_territory_ids = df_avg_due_top10['territory_id'].tolist()
df_filtered_top10_territories = df[df['territory_id'].isin(top10_territory_ids)].copy()

# Map the territory_id in the filtered DataFrame to numerical IDs
df_filtered_top10_territories['territory_numerical_id'] = df_filtered_top10_territories['territory_id'].map(territory_id_mapping)

# Calculate the count of orders for each numerical territory ID
territory_counts = df_filtered_top10_territories['territory_numerical_id'].value_counts().reset_index()
territory_counts.columns = ['territory_numerical_id', 'count']
territory_counts = territory_counts.sort_values(by='territory_numerical_id') # Sort by numerical ID for consistent line chart order

# Counts
lines = alt.Chart(territory_counts).mark_line(point=True, color='steelblue').encode(
    x=alt.X('territory_numerical_id:O', title='Territory ID (Numerical)'),
    y=alt.Y('count', axis=alt.Axis(title='Total Count of Orders', titleColor='steelblue')),
    tooltip=['territory_numerical_id', 'count']
)

# Layer the two charts and resolve scales for independent Y-axes
combined_layered_chart = alt.layer(bars, lines).resolve_scale(
    y='independent'
).properties(
    title='Average Total Due and Order Counts by Territory ID (Numerical Labels)'
)

combined_layered_chart.display()
```

```python
feature_2_insights = """
The territory_id has 10 unique values, and a chart showing the average total_due for each of these territories has been generated.
To simplify analysis and plotting, these UUIDs have been mapped to numerical IDs from 1 to 10.
The top 3 territories with the highest average total due are represented by numerical IDs 1, 2, and 3.

In terms of order count, the top 3 territories are represented by numerical IDs 10, 5, and 6 respectively.
This finding reveals that order volume and average total due show opposite trends.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```
