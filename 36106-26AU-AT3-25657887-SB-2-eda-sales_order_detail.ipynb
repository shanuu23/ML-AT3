### B.1 Explore Dataset

```python
df.head()
```

```python
# Basic dataset information
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame(name="data_type"))

print("\nSummary statistics for numerical columns:")
display(df.describe(include=[np.number]))

print("\nSummary statistics for categorical columns:")
display(df.describe(include=["object"]))
```

```python
# Missing value analysis
missing_summary = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": (df.isna().mean() * 100).round(2),
    "non_missing_count": df.notna().sum()
}).sort_values(by="missing_percentage", ascending=False)

missing_summary
```

```python
# Duplicate and uniqueness checks
duplicate_summary = pd.DataFrame({
    "metric": [
        "Total rows",
        "Total columns",
        "Fully duplicated rows",
        "Unique sales_order_id values",
        "Unique sales_order_detail_id values",
        "Unique product_id values",
        "Unique special_offer_id values"
    ],
    "value": [
        len(df),
        df.shape[1],
        df.duplicated().sum(),
        df["sales_order_id"].nunique(),
        df["sales_order_detail_id"].nunique(),
        df["product_id"].nunique(),
        df["special_offer_id"].nunique()
    ]
})

duplicate_summary
```

```python
# Check whether sales_order_detail_id is unique
detail_id_check = pd.DataFrame({
    "metric": [
        "Number of rows",
        "Unique sales_order_detail_id values",
        "Duplicate sales_order_detail_id values"
    ],
    "value": [
        len(df),
        df["sales_order_detail_id"].nunique(),
        len(df) - df["sales_order_detail_id"].nunique()
    ]
})

detail_id_check
```

```python
# Visualise missing percentage by column
missing_chart_data = missing_summary.reset_index().rename(columns={"index": "column"})

alt.Chart(missing_chart_data).mark_bar().encode(
    x=alt.X("missing_percentage:Q", title="Missing Percentage (%)"),
    y=alt.Y("column:N", sort="-x", title="Column"),
    tooltip=["column", "missing_count", "missing_percentage"]
).properties(
    title="Missing Values by Column",
    width=700,
    height=300
)
```

```python
# Correlation between numerical features and line_total
numeric_cols = ["order_quantity", "unit_price", "unit_price_discount", "line_total"]

correlation_table = (
    df[numeric_cols]
    .corr()
    .round(3)
)

correlation_table
```

```python
# Correlation with target variable
target_correlation = (
    df[numeric_cols]
    .corr()["line_total"]
    .sort_values(ascending=False)
    .reset_index()
)

target_correlation.columns = ["feature", "correlation_with_line_total"]

target_correlation
```

```python
dataset_insights = """
The sales_order_detail dataset contains 42,100 rows and 8 columns. Each row represents a sales order line item, meaning one sales order can contain multiple product lines. The dataset includes sales_order_id, sales_order_detail_id, order_quantity, product_id, special_offer_id, unit_price, unit_price_discount, and line_total.
This dataset is highly relevant to the regression use case because it contains line_total, which is a continuous numerical value and can be used as a target variable. The dataset can support a business use case such as predicting the sales amount of an order line item or, after aggregation, predicting the total sales value of a full order.
The dataset is clean in terms of missing values because none of the columns contain missing data. There are also no fully duplicated rows, and sales_order_detail_id is unique for every row. This means the dataset requires less basic cleaning than the person dataset.
The dataset contains 15,357 unique sales orders, 42,100 unique sales order detail records, 254 unique products, and 10 unique special offers. This shows that the data contains multiple products and offer types across many orders.
"""
```

### B.2 Explore Feature of Interest `line_total`

```python
# Summary statistics for line_total
line_total_summary = df["line_total"].describe().to_frame(name="line_total")

line_total_summary
```

```python
# Additional target distribution statistics
line_total_stats = pd.DataFrame({
    "metric": [
        "Minimum",
        "25th percentile",
        "Median",
        "Mean",
        "75th percentile",
        "Maximum",
        "Standard deviation",
        "Skewness"
    ],
    "value": [
        df["line_total"].min(),
        df["line_total"].quantile(0.25),
        df["line_total"].median(),
        df["line_total"].mean(),
        df["line_total"].quantile(0.75),
        df["line_total"].max(),
        df["line_total"].std(),
        df["line_total"].skew()
    ]
})

line_total_stats
```

```python
# Histogram of line_total
alt.Chart(df).mark_bar().encode(
    x=alt.X(
        "line_total:Q",
        bin=alt.Bin(maxbins=60),
        title="Line Total"
    ),
    y=alt.Y("count():Q", title="Number of Order Lines"),
    tooltip=["count()"]
).properties(
    title="Distribution of Line Total",
    width=700,
    height=350
)
```

```python
# Because line_total is right-skewed, log transformation helps visualise it better
line_total_log_data = df.copy()
line_total_log_data["log_line_total"] = np.log1p(line_total_log_data["line_total"])

alt.Chart(line_total_log_data).mark_bar().encode(
    x=alt.X(
        "log_line_total:Q",
        bin=alt.Bin(maxbins=60),
        title="Log Transformed Line Total"
    ),
    y=alt.Y("count():Q", title="Number of Order Lines"),
    tooltip=["count()"]
).properties(
    title="Distribution of Log-Transformed Line Total",
    width=700,
    height=350
)
```

```python
feature_1_insights = """
The line_total feature is the most important variable in this dataset for the regression use case because it can be used as the target variable. It represents the total value of each sales order line item.
The distribution of line_total is highly right-skewed. Most order lines have relatively low sales values, while a smaller number of line items have very high values. This is shown by the difference between the median and the mean, where the mean is much higher than the median. This suggests that expensive products or high-quantity purchases create large sales values.
This skewness is important for regression modelling.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_1_insights', value=feature_1_insights)
```

### B.3 Explore Feature of Interest `order_quantity`

```python
order_quantity_summary = df["order_quantity"].describe().to_frame(name="order_quantity")

order_quantity_summary
```

```python
# Frequency table for order_quantity
order_quantity_counts = (
    df["order_quantity"]
    .value_counts()
    .reset_index()
)

order_quantity_counts.columns = ["order_quantity", "count"]
order_quantity_counts["percentage"] = (
    order_quantity_counts["count"] / len(df) * 100
).round(2)

order_quantity_counts = order_quantity_counts.sort_values("order_quantity")

order_quantity_counts.head(20)
```

```python
# Bar chart of order quantity distribution
alt.Chart(order_quantity_counts).mark_bar().encode(
    x=alt.X("order_quantity:O", title="Order Quantity"),
    y=alt.Y("count:Q", title="Number of Order Lines"),
    tooltip=["order_quantity", "count", "percentage"]
).properties(
    title="Distribution of Order Quantity",
    width=700,
    height=350
)
```

```python
# Relationship between order_quantity and line_total
alt.Chart(df).mark_circle(opacity=0.35).encode(
    x=alt.X("order_quantity:Q", title="Order Quantity"),
    y=alt.Y("line_total:Q", title="Line Total"),
    tooltip=["order_quantity", "unit_price", "unit_price_discount", "line_total"]
).properties(
    title="Relationship Between Order Quantity and Line Total",
    width=700,
    height=400
)
```

```python
# Average line_total by order_quantity
avg_line_total_by_quantity = (
    df.groupby("order_quantity")
    .agg(
        average_line_total=("line_total", "mean"),
        median_line_total=("line_total", "median"),
        count=("line_total", "count")
    )
    .reset_index()
)

avg_line_total_by_quantity.head(20)
```

```python
feature_2_insights = """
The order_quantity feature shows how many units of a product were purchased in each sales order line. The distribution is highly concentrated at low quantities, especially quantity 1. This suggests that most order lines involve single-item purchases, while larger quantity purchases are less common.
Order quantity is important for the regression use case because it has a direct relationship with line_total. In general, higher quantities should increase the total value of a line item. The scatter plot and average line total by quantity help show that line_total often increases as order_quantity increases, although the relationship also depends heavily on unit_price.
"""
```

```python
# Do not modify this code
print_tile(size="h3", key='feature_2_insights', value=feature_2_insights)
```

### B.4 Explore Feature of Interest `unit_price`

```python
# Summary statistics for unit_price
unit_price_summary = df["unit_price"].describe().to_frame(name="unit_price")

unit_price_summary
```

```python
# Additional statistics for unit_price
unit_price_stats = pd.DataFrame({
    "metric": [
        "Minimum",
        "25th percentile",
        "Median",
        "Mean",
        "75th percentile",
        "Maximum",
        "Standard deviation",
        "Skewness"
    ],
    "value": [
        df["unit_price"].min(),
        df["unit_price"].quantile(0.25),
        df["unit_price"].median(),
        df["unit_price"].mean(),
        df["unit_price"].quantile(0.75),
        df["unit_price"].max(),
        df["unit_price"].std(),
        df["unit_price"].skew()
    ]
})

unit_price_stats
```

```python
# Top 10 most common unit prices
unit_price_counts = (
    df["unit_price"]
    .value_counts()
    .reset_index()
)

unit_price_counts.columns = ["unit_price", "count"]
unit_price_counts = unit_price_counts.sort_values("count", ascending=False)

unit_price_counts.head(10)
```

```python
# Histogram of unit_price
alt.Chart(df).mark_bar().encode(
    x=alt.X(
        "unit_price:Q",
        bin=alt.Bin(maxbins=60),
        title="Unit Price"
    ),
    y=alt.Y("count():Q", title="Number of Order Lines"),
    tooltip=["count()"]
).properties(
    title="Distribution of Unit Price",
    width=700,
    height=350
)
```

```python
# Relationship between unit_price and line_total
alt.Chart(df).mark_circle(opacity=0.35).encode(
    x=alt.X("unit_price:Q", title="Unit Price"),
    y=alt.Y("line_total:Q", title="Line Total"),
    tooltip=[
        "order_quantity",
        "unit_price",
        "unit_price_discount",
        "line_total"
    ]
).properties(
    title="Relationship Between Unit Price and Line Total",
    width=700,
    height=400
)
```

```python
# Average line_total by unit_price bands

unit_price_band_data = df.copy()

unit_price_band_data["unit_price_band"] = pd.qcut(
    unit_price_band_data["unit_price"],
    q=5,
    duplicates="drop"
)

avg_line_total_by_price_band = (
    unit_price_band_data
    .groupby("unit_price_band", observed=True)
    .agg(
        average_line_total=("line_total", "mean"),
        median_line_total=("line_total", "median"),
        count=("line_total", "count")
    )
    .reset_index()
)

avg_line_total_by_price_band["unit_price_band"] = (
    avg_line_total_by_price_band["unit_price_band"].astype(str)
)

avg_line_total_by_price_band
```

```python
feature_n_insights = """
The unit_price feature represents the price of each product unit in a sales order line. It is one of the most important numerical variables in the sales_order_detail dataset because it has a strong relationship with line_total.
The distribution of unit_price is right-skewed. Most order lines have relatively low unit prices, while a smaller number of products have much higher prices. This suggests that the retailer sells products across different price ranges, including lower-value products and high-value products.
The scatter plot between unit_price and line_total shows that line_total generally increases as unit_price increases. However, the relationship is also affected by order_quantity and discount. A high unit price with a low quantity may still produce a smaller line total than a lower unit price with a large quantity.
For the regression use case, unit_price is expected to be a strong predictor of sales amount.
"""
```
